# String

### Íncide e tamanho de String

```python
email = "joao@gmail.com"
nome = "João Breno"

# para pegar o tamanho de um texto, é só colocar o len()
# pegar um item dentro de uma string é só colocar o texto ou a variável e [indice]

print(len(nome))
print(email[4])
```

### Índice Negativo e Pedaço de String

```python
email = "joao@gmail.com"
nome = "João Breno"

print(nome[-2])
print(email[:-10])
print(email[5:10])
```

### Operações com String

```python
faturamento = 1000
custo = 500
lucro = faturamento - custo

print(f'O faturamento da loja foi de {str(faturamento)}')

print('@' in 'joao@gmail.com')
print('@' in 'joao.gmail.com')
```

### Métodos de String - Apresentação

```python
texto = 'joão breno souza '

print(texto.upper())
print(texto.lower())
print(texto.capitalize())
print(texto.title())
print(texto.count('o'))
print(len(texto))
print(len(texto.strip()))
print(texto.casefold())
print(texto.endswith(' '))
print(texto.startswith('j'))
print(texto.find('ã'))
print(texto.isalnum())
print(texto.isalpha())
print(texto.isnumeric())
print(texto.replace(' ', ''))
print(texto.split(' '))
```

### Formatação de números

```python
email = 'joao@gmail.com'
custo = 5000
faturamento = 2700
lucro = faturamento - custo
margem = lucro / faturamento
lucro_texto = 'R${:_.2f}'.format(lucro)

print(f'Meu email é {email:<30}?')
print(f'Faturamento foi {faturamento:+} e lucro foi {lucro:+}')
print(f'Faturamento foi {faturamento:+,} e lucro foi {lucro:+,}')
print(f'Faturamento foi {faturamento:.2f} e lucro foi {lucro:.2f}')
print(f'Margem de lucro foi de {margem:.2%}')
print(f'Faturamento foi R${faturamento:,.2f} e lucro foi R${lucro:,.2f}')
print(lucro_texto.replace('.', ',').replace('_', '.'))
```
