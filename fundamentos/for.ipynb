# For

### Estrutura de Repetição For

```python
produtos = ['tv', 'celular', 'tablet', 'mouse', 'teclado', 'geladeira']

for i in range(len(produtos)):
    print(produtos[i])
```

### For each - Percorrer cada item de uma lista

```python
produtos = ['tv', 'celular', 'tablet', 'mouse', 'teclado', 'geladeira']
for item in produtos:
    print(item)
```

### Enumerate - For com item e índice

```python
produtos = ['tv', 'celular', 'tablet', 'mouse', 'teclado', 'geladeira']

for i, produto in enumerate(produtos):
    print(f'{i} é o produto {produto}')
```
