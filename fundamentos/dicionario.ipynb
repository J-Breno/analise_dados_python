# Dicionário

### Pegar item

```python
mais_vendidos = {'tecnologia': 'iphone', 'refrigeracao': 'ar consul', 'livros': '48 leis do poder'}

print(mais_vendidos['tecnologia'])
print(mais_vendidos.get('refrigeracao'))
```

### Adicionar, Modificar e Remover Itens do Dicionário

```python
mais_vendidos = {'tecnologia': 'iphone', 'refrigeracao': 'ar consul', 'livros': '48 leis do poder'}
mais_vendidos['curso'] = 'Python'
mais_vendidos.update({'treinamento': 'sql'})
print(mais_vendidos)
del mais_vendidos['curso']
print(mais_vendidos)
```
