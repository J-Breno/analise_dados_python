# IF

### If - Estrutura básica

```python
meta = 5000
qtde_vendida = 30000

if qtde_vendida > meta:
    print('Batemos a meta de vendas de Iphone, vendemos {} unidades'.format(qtde_vendida))
else:
    print('Infelizmente não batemos a meta, pois a meta seria {} unidades e vendemos {} unidades'.format(meta, qtde_vendida))
```

### If - Blocos e identação

```python
meta = 0.05
taxa = 0
rendimento = 0.25

if rendimento > meta:
    if rendimento > 0.20:
        taxa = 0.04
        print(f'A taxa foi de {taxa}')
    else:
        taxa = 0.02
        print(f'A taxa foi de {taxa}')
else:
    taxa = 0
    print(f'A taxa foi de {taxa}')
```

### Elif

```python
meta = 20000
vendas = 25000

if vendas < meta:
    print('Não ganhou bônus')
elif vendas > (meta * 2):
    bonus = 0.07 * vendas
    print(f'Ganhou {bonus} de bônus')
else:
    bonus = 0.03 * vendas
    print(f'Ganhou {bonus} de bônus')
```
