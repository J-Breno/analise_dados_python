# Listas

### Descobrindo o Índice de um item na lista

```python
produtos = ['tv', 'celular', 'tablet', 'mouse', 'teclado', 'geladeira']
estoque = [100, 150, 200, 250, 300, 350]

i = produtos.index('tablet')
print(i)
```

### Adicionar e Remover itens da Lista

```python
produtos = ['tv', 'celular', 'tablet', 'mouse', 'teclado', 'geladeira']
# pop você pode excluir um item de uma lista com um indice e você pode armazena-lo em algum local
# existe também o remove
# para adicionar é o append

produtos.append('iphone')
print(produtos)
produto_iphone = produtos.pop(-1)
print(produto_iphone)
print(produtos)
produtos.remove('teclado')
print(produtos)
```

### Juntar Listas e Ordenar

```python
# O método extend muda sua lista original
# para ordenar as listas pode usar o sort ( tabela ASCII)


produtos = ['tv', 'celular', 'tablet', 'mouse', 'teclado', 'geladeira']
novos_produtos = ['iphone', 'ipad']


print(produtos)
print(novos_produtos)
produtos.extend(novos_produtos)
print(produtos)
```
